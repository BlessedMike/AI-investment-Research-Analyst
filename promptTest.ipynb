```python
# Convert stock data to a readable format for analysis
stock_summary = stock_filtered.head(20).to_string()

completion = client.chat.completions.create(
  model="nvidia/nvidia-nemotron-nano-9b-v2",
  messages=[
    {"role":"system","content":"/think"}, 
    {"role":"user","content":f"Here is recent stock market data from a CSV file:\n\n{stock_summary}\n\nAnalyze this data through multiple lenses: identify key trends and patterns in the price movements, research relevant news articles and market sentiment for each asset, and perform comparative analysis between different stocks and commodities. Based on your comprehensive analysis, provide a clear BUY or DON'T BUY recommendation for the assets. Your response must include exactly 3 sentences that explain your reasoning and justify your recommendation. no more no less, nothing before that either and it must be in a paragraph format."}
  ],
  temperature=0.3,
  top_p=0.9,
  max_tokens=2048,
  frequency_penalty=0.1,
  presence_penalty=0,
  stream=True,
  extra_body={
    "min_thinking_tokens": 1024,
    "max_thinking_tokens": 2048
  }
)
```

```python
for chunk in completion:
  reasoning = getattr(chunk.choices[0].delta, "reasoning_content", None)
  if reasoning:
    print(reasoning, end="")
  if chunk.choices[0].delta.content is not None:
    print(chunk.choices[0].delta.content, end="")
```
